```python
def add(first, second):
    print(first + second)
```

```python
def sub(first, second):
    print(first - second)
```

```python
def calculate(first, second, third=0):
    if(third==0):
        return add(first, second)
    else:
        return sub(first, second)
```

```python
calculate(2,5,6)
```

```python
class rightAngledQuadrilateral:
    def __init__(self, side, breadth=0):
        self.side = side
        self.breadth = breadth

    def area(self):
        if(self.breadth == 0):
            return self.areaOfSquare(self.side)
        else:
            return self.areaOfRectangle(self.side, self.breadth)

    def areaOfSquare(self,side):
        return self.side*self.side

    def areaOfRectangle(self,side, breadth):
        return self.side*self.breadth
```

```python
square = rightAngledQuadrilateral(25)
rectangle = rightAngledQuadrilateral(5,4)
```

```python
square.area()
```

```python
rectangle.area()
```

```python
class fruit:
    def sweet(self):
        print("All fruits are not sweet")

class mango(fruit):
    def sweet(self):
        print("Some mangoes are sweet")

class cherry(fruit):
    def sweet(self):
        # print("Cherry is sweet fruit")
        super().sweet()
```

```python
a1 = mango()
b1 = cherry()
```

```python
a1.sweet()
```

```python
b1.sweet()
```

```python
class vehicle:
    def petrol(self):
        print("All vehicles are new for next 15 years.")

    def diesel(self):
        print("All vehicles are new for next 10 years.")

class hundayi(vehicle):
    def petrol(self):
        print("If its hundayi and Petrol then its Venue")

    def diesel(self):
        print("If its hundayi and Petrol then its Creta")

class honda(vehicle):
    def petrol(self):
        # print("If its Honda and Petrol then its city")
        super().petrol()

    def diesel(self):
        print("If its Honda and Diesel then its civic")
```

```python
a1 = hundayi()
b1 = honda()
```

```python
a1.petrol()
```

```python
b1.petrol()
```

```python
b1.diesel()
```
