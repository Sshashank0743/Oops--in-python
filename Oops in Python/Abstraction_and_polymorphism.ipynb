```python
class rightAngledQuadrilateral:
    def __init__(self, side, breadth=0):
        self.side = side
        self.breadth = breadth

    def area(self):
        if(self.breadth == 0):
            return self.areaOfSquare(self.side)
        else:
            return self.areaOfRectangle(self.side, self.breadth)

    def areaOfSquare(self,side):
        return self.side*self.side

    def areaOfRectangle(self,side, breadth):
        return self.side*self.breadth
```

```python
square = rightAngledQuadrilateral(25)
rectangle = rightAngledQuadrilateral(5,4)
```

```python
square.area()
```

```python
class Book:
    def __init__(self,title,author,ISBN,genre):
        self.title = title
        self.author = author
        self.ISBN = ISBN
        self.genre = genre
        self.availability = True
        
    # display_book_info(), check_availability(), lend_book(), return_book()
    def display_book_info(self):
        print(f"{self.title},{self.author},{self.ISBN},{self.genre}")
        
    def check_availability(self):
        return self.availability
    
    def lend_book(self):
        if(self.availability):
            self.availability = False
            print("Book Lent Successfully")
        else:
            print("Book not available")
            
    def return_book(self):
        if( self.availability):
            print("Book is alreadty with us")
        else:
            self.availability = True
            print("Book returned successfuly")
```

```python
class Member:
    def __init__(self,name,member_id):
        self.name = name
        self.member_id=member_id
        self.books_borrowed = []
        
    def borrow_book(self,book):
        if(book.check_availability()):
            self.books_borrowed.append(book)
            book.lend_book()
        else:
            print("Book not available")
    
    def return_book(self,book):
        if book in self.books_borrowed:
            self.books_borrowed.remove(book)
            book.return_book()
        else:
            print("The member has not borrowed this book")
```

```python
class Product:
    def __init__(self, name, price, quantity_available):
        self.name = name
        self.price = price
        self.quantity_available = quantity_available
    
    # Method to check availability of the product
    def check_availability(self, quantity):
        return self.quantity_available >= quantity
    
    # Method to update the quantity of the product
    def update_quantity(self, quantity):
        if quantity <= self.quantity_available:
            self.quantity_available -= quantity
            return True
        return False

    def __str__(self):
        return (f"{self.name} - ${self.price} (Available: {self.quantity_available})")


class Customer:
    def __init__(self, name, email, address):
        self.name = name
        self.email = email
        self.address = address
        self.cart = []
        self.order = None
    
    # Method to add a product to the customer's cart
    def add_to_cart(self, product, quantity):
        if product.check_availability(quantity):
            self.cart.append((product, quantity))
            print(f"{quantity} x {product.name} added to {self.name}'s cart.")
        else:
            print(f"Not enough stock for {product.name}!")

    def display_cart(self):
        if not self.cart:
            print(f"{self.name}'s cart is empty.")
        else:
            print(f"{self.name}'s Cart:")
            for product, qty in self.cart:
                print(f"- {product.name} x {qty} @ ${product.price} each")

    # Method to place an order for the products in the cart
    def place_order(self):
        if not self.cart:
            print("Cart is empty. Cannot place order.")
            return None
        
        total_price = sum(product.price * qty for product, qty in self.cart)
        order_id = f"ORD{hash(self.name + self.email) % 10000}"
        self.order = Order(order_id, self.cart, total_price)
        
        # Update product quantities
        for product, qty in self.cart:
            product.update_quantity(qty)
        
        print(f"Order {self.order.order_id} placed successfully for {self.name}.")
        self.cart.clear()  # Empty cart after order
        return self.order
    
    # Method to calculate the total price of the order
    def calculate_order_total(self):
        if self.order:
            return self.order.total_price
        else:
            print("No order has been placed yet.")
            return 0


class Order:
    def __init__(self, order_id, products_ordered, total_price):
        self.order_id = order_id
        self.products_ordered = products_ordered
        self.total_price = total_price
    
    # Method to calculate the total price of the order
    def calculate_total_price(self):
        return sum(product.price * qty for product, qty in self.products_ordered)


class OnlineShoppingSystem:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def display_available_products(self):
        print("Available Products:")
        for product in self.products:
            print(product)


# Test Cases

# Create online shopping system and products
shopping_system = OnlineShoppingSystem()
product1 = Product("Laptop", 1000, 5)
product2 = Product("Headphones", 50, 10)

shopping_system.add_product(product1)
shopping_system.add_product(product2)

# Test Case 1: Display available products
shopping_system.display_available_products()

# Test Case 2: Create customer and add products to cart
customer1 = Customer("Alice", "alice@example.com", "123 Main St")
customer1.add_to_cart(product1, 2)
customer1.add_to_cart(product2, 1)

# Test Case 3: Display cart contents
customer1.display_cart()

# Test Case 4: Place order
customer1.place_order()

# Test Case 5: Calculate total price of order
order_total = customer1.calculate_order_total()
print("Order Total:", order_total)
```
